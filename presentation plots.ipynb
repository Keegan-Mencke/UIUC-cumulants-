```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os

# ============================================================
# Parameters
# ============================================================

v2_values = [0.05, 0.09]
n_values = [100, 200]
scale = 1.0

# ============================================================
# Read data and calculate percent error
# ============================================================

data = {}
dx = {0.05: -0.04, 0.09: 0.04}
for v2 in v2_values:
    vtag = "v205" if v2 == 0.05 else "v209"

    for n in n_values:
        df = pd.read_csv(f"pure_jets_{vtag}_n{n}.csv")

        dn4 = df["dn4e4"].to_numpy()
        err = df["dn4stde4"].to_numpy()
        pt = df["pt_bins"].to_numpy()

        truth = -v2**4

        # Percent error relative to truth
        percent_error = 100 * np.abs(dn4 - truth) / np.abs(truth)

        # Propagate statistical error into percent error
        percent_error_err = 100 * err / np.abs(truth)

        data[(v2, n)] = (pt+dx[v2], percent_error, percent_error_err)


# ============================================================
# Plot
# ============================================================

plt.rcParams.update({
    "font.size": 20, "axes.labelsize": 18, "axes.titlesize": 21,
    "legend.fontsize": 14, "xtick.labelsize": 15, "ytick.labelsize": 15,
    "xtick.direction": "in", "ytick.direction": "in",
    "xtick.top": True, "ytick.right": True,
    "xtick.minor.visible": True, "ytick.minor.visible": True,
    "axes.linewidth": 1.2, "font.family": "serif"
})

fig, ax = plt.subplots(figsize=(8, 6))

styles = {
    (0.05, 100): ('^', 'blue', 'v2 = 0.05, N = 100'),
    (0.05, 200): ('o', 'blue', 'v2 = 0.05, N = 200'),
    (0.09, 100): ('^', 'red',  'v2 = 0.09, N = 100'),
    (0.09, 200): ('o', 'red',  'v2 = 0.09, N = 200'),
}

for key, (marker, color, label) in styles.items():
    pt, pe, pe_err = data[key]

    ax.errorbar(pt, pe, yerr=pe_err, marker=marker, color=color, mfc="none",
                linestyle='none', capsize=3, label=label)

ax.axhline(0, color='black', linewidth=1)
ax.grid(axis='y', which='major', linestyle='--', linewidth=0.6, alpha=0.3)

ax.set_xlim(0, 7)
ax.set_xlabel(r"$p_t$ (GeV)")
ax.set_ylabel("Percent error (%)")
ax.set_title(r"$d_2\{4\}$ 4-subevent percent error")
ax.legend(frameon=False)

fig.tight_layout()


# ============================================================
# Save
# ============================================================

OUTDIR = "plots"
os.makedirs(OUTDIR, exist_ok=True)

fig.savefig(os.path.join(OUTDIR, "dn4_e4_percent_error.png"), bbox_inches="tight")
fig.savefig(os.path.join(OUTDIR, "dn4_e4_percent_error.pdf"), bbox_inches="tight")

print("Saved dn4_e4_percent_error.png / .pdf to plots/")

plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd
import os 

# ============================================================
# Parameters
# ============================================================

datasets = {
    0.05: [50, 110, 1000],
    0.09: [110, 1000, 10000]
}

dx = {0.05: -0.04, 0.09: 0.04}

# ============================================================
# Read data and calculate percent error
# ============================================================

data = {}

for v2, n_values in datasets.items():
    vtag = "v205" if v2 == 0.05 else "v209"

    for n in n_values:
        df = pd.read_csv(f"pp_isomerge_{vtag}_{n}Nparticles.csv")

        dn4 = df["dn4e4"].to_numpy()
        err = df["dn4stde4"].to_numpy()
        pt = df["pt_bins"].to_numpy()

        truth = -v2**4
        percent_error = 100 * np.abs(dn4 - truth) / np.abs(truth)
        percent_error_err = 100 * err / np.abs(truth)

        data[(v2, n)] = (pt + dx[v2], percent_error, percent_error_err)

# ============================================================
# Plot style
# ============================================================

plt.rcParams.update({
    "font.size": 20, "axes.labelsize": 18, "axes.titlesize": 21,
    "legend.fontsize": 14, "xtick.labelsize": 15, "ytick.labelsize": 15,
    "xtick.direction": "in", "ytick.direction": "in",
    "xtick.top": True, "ytick.right": True,
    "xtick.minor.visible": True, "ytick.minor.visible": True,
    "axes.linewidth": 1.2, "lines.markersize": 8,
    "lines.markeredgewidth": 1.4, "figure.dpi": 120,
    "savefig.dpi": 300, "font.family": "serif"
})

# ============================================================
# Plot
# ============================================================

fig, ax = plt.subplots(figsize=(9, 7))

styles = {
    (0.05, 50):   ('^', 'blue'),
    (0.05, 110):  ('o', 'blue'),
    (0.05, 1000): ('s', 'blue'),
    (0.09, 110):  ('^', 'red'),
    (0.09, 1000): ('o', 'red'),
    (0.09, 10000):('s', 'red'),
}

for (v2, n), (marker, color) in styles.items():
    pt, pe, pe_err = data[(v2, n)]
    ax.errorbar(pt, pe, yerr=pe_err, marker=marker, color=color,
                mfc="none", linestyle="none", capsize=3,
                label=fr"$v_2={v2},\ N={n}$")

ax.axhline(0, color='black', linewidth=1)
ax.grid(axis='y', which='major', linestyle='--', linewidth=0.6, alpha=0.3)

ax.set_xlim(0, 7)
ax.set_xlabel(r"$p_t$ (GeV)")
ax.set_ylabel("Percent error (%)")
ax.set_title(r"$d_2\{4\}$ 4-subevent percent error — pp")
ax.legend(frameon=False)

fig.tight_layout()

# ============================================================
# Save
# ============================================================

OUTDIR = "plots"
os.makedirs(OUTDIR, exist_ok=True)

fig.savefig(os.path.join(OUTDIR, "pp_dn4_e4_percent_error.png"), bbox_inches="tight")
fig.savefig(os.path.join(OUTDIR, "pp_dn4_e4_percent_error.pdf"), bbox_inches="tight")

print("Saved pp_dn4_e4_percent_error.png / .pdf to plots/")

plt.show()
```
